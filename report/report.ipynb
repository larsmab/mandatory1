 ## 1.2.3.

 We have the wave equation

 $$
\frac {\partial ^2 u}{\partial t^2}=c^2\nabla ^2 u,
 $$

 and are given the candidate solution

 $$
u(t,x,y)=e^{\imath(k_xx+k_yy-\omega t)}.
 $$

We find the second partial derivatives

$$
\frac{\partial^2 u}{\partial t^2} = (-\imath\omega)^2 u = -\omega^2 u, \qquad
\frac{\partial^2 u}{\partial x^2} = (\imath k_x)^2 u = -k_x^2 u, \qquad
\frac{\partial^2 u}{\partial y^2} = (\imath k_y)^2 u = -k_y^2 u,$$

and have

$$
\nabla ^ 2 u = \frac{\partial^2 u}{\partial x^2} + \frac{\partial^2 u}{\partial y^2} = -k_x^2 u -k_y^2 u.
$$


Inserting into the wave equation gives

$$
\begin{aligned}
-\omega^2 u &= -c^2(k_x^2 u +k_y^2 u)\\
\omega^2 &= c^2(k_x^2+k_y^2)
\end{aligned}.
$$
Since $|u| = 1 \neq 0$ we can divide by $-u$.
Hence, the candidate solution is correct iff

$$
\omega =\pm c \sqrt{k_x^2+k_y^2}
$$

## 1.2.4.


We assume $m_x = m_y$, so that $k_x = k_y = k$, and $C = c\Delta t/h = 1/\sqrt{2}$. We want to insert the discrete wave
$$
u^n_{ij} = e^{\imath(kh(i+j) - \tilde\omega n \Delta t)}
$$

into the scheme:
$$
\frac{u_{i,j}^{n+1} - 2u_{i,j}^{n} + u_{i,j}^{n-1}}{\Delta t^2} = c^2 \left(
\frac{u_{i+1,j}^{n} - 2u_{i,j}^{n} + u_{i-1,j}^{n}}{h^2} + \frac{u_{i,j+1}^{n} - 2u_{i,j}^{n} + u_{i,j-1}^{n}}{h^2} \right ) \quad \text{(1)}.
$$


Multiplying the scheme (1) by $\Delta t^2$ and using $C^2 = c^2\Delta t^2/h^2 = \tfrac{1}{2}$ gives
$$
u^{n+1}_{ij} - 2u^n_{ij} + u^{n-1}_{ij} = \tfrac{1}{2}\left(u^n_{i+1,j} - 2u^n_{ij} + u^n_{i-1,j} + u^n_{i,j+1} - 2u^n_{ij} + u^n_{i,j-1}\right).
$$


A useful piece of information is that shifted indices only add a constant to the exponent.
$$
u^{n+1}_{ij} = e^{\imath(kh(i+j) - \tilde\omega(n+1)\Delta t)} = e^{\imath(kh(i+j) - \tilde\omega n\Delta t)}\, e^{-\imath\tilde\omega\Delta t} = e^{-\imath\tilde\omega\Delta t}\,u^n_{ij},
$$
$$
u^n_{i+1,j} = e^{\imath(kh(i+1+j) - \tilde\omega n\Delta t)} = e^{\imath(kh(i+j) - \tilde\omega n\Delta t)}\, e^{\imath kh} = e^{\imath kh}\,u^n_{ij}.
$$
Similar derivations apply for negative shifting of $n$ and $i$, and shifting of $j$.

We insert $u^n_{ij} = e^{\imath(kh(i+j) - \tilde\omega n\Delta t)}$ and divide both sides by $e^{\imath(kh(i+j) - \tilde\omega n\Delta t)}$:

$$
e^{-\imath\tilde\omega\Delta t} - 2 + e^{\imath\tilde\omega\Delta t} = \tfrac{1}{2}\left(e^{\imath kh} - 2 + e^{-\imath kh} + e^{\imath kh} - 2 + e^{-\imath kh}\right) = e^{\imath kh} - 2 + e^{-\imath kh},
$$
where we used the indexing trick explained above.

Using $e^{\imath\theta} + e^{-\imath\theta} = 2\cos\theta$, we get:
$$
2\cos (\tilde \omega \Delta t) -2 = 2\cos(kh) -2,
$$
and subsequently,
$$
\cos(\tilde\omega\Delta t) = \cos(kh) \quad\Rightarrow\quad \tilde\omega\Delta t = \pm kh + 2\pi z, \quad z \in \mathbb{Z}.
$$

Since $n$ is an integer, adding $2\pi z$ to $\tilde\omega\Delta t$ does not change $e^{-\imath\tilde\omega n\Delta t}$, so we choose $z=0$. Taking $\tilde\omega\Delta t = \pm kh$, and using that $C=\frac {c\Delta t}h \left (\implies  \frac h {\Delta t}=\frac c C\right)$ and that $C=\frac 1 {\sqrt{2}}$:

$$
\tilde\omega = \pm\frac{kh}{\Delta t} = \pm\frac{kc}{C} = \pm\sqrt{2}\,ck.
$$
The exact value is, from 1.2.3, $\omega = \pm c\sqrt{k^2 + k^2} = \pm\sqrt{2}\,ck$. Choosing the same sign for both frequencies, we get $\tilde\omega = \omega$.
